```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
```

```python
# Use a pipeline as a high-level helper
from transformers import pipeline

pipe_finbert = pipeline("text-classification", model="ProsusAI/finbert")
```

```python
pipe_finbert("The company announced positive earnings results on Tuesday.", top_k=None, truncate=True)
```

```python
# Use a pipeline as a high-level helper
from transformers import pipeline

pipe_twitter = pipeline(
    "text-classification", 
    model="cardiffnlp/twitter-roberta-base-sentiment-latest",
    device=0, # gpu, to switch to cpu, use device = -1
    batch_size=64
)
```

```python
pipe_finbert("Gold will be priceless soon", top_k=None, truncation=True)
```

```python
pipe_twitter("Gold will be priceless soon", top_k=None, truncation=True)
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
models=[
  ('ProsusAI/finbert','/opt/models/finbert'),
  ('cardiffnlp/twitter-roberta-base-sentiment-latest','/opt/models/twitter_roberta_sentiment')
]
for name, path in models:
  AutoTokenizer.from_pretrained(name, cache_dir=path)
  AutoModelForSequenceClassification.from_pretrained(name, cache_dir=path)
```
